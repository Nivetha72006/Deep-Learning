```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.layers import LSTM, Embedding, Input, Dense, Bidirectional
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
```

```python
vocab_size = 10000
imdb=keras.datasets.imdb
(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=vocab_size
)
```

```python
inputs = Input(shape=(200,))
# Add an Embedding layer to convert integer sequences to dense vectors
embedding_dim = 128 # You can choose a different embedding dimension
x = Embedding(vocab_size, embedding_dim)(inputs)
lstm_layer =  Bidirectional(LSTM(128))(x)
outputs = Dense(1,activation='sigmoid')(lstm_layer)
model = Model(inputs, outputs)
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
x_train_padded = pad_sequences(x_train, maxlen=200)
x_test_padded = pad_sequences(x_test, maxlen=200)

history = model.fit(x_train_padded,y_train,validation_split=0.1,epochs=5)
```

```python
y_pred = model.predict(x_test_padded)

y_pred1 = (y_pred > 0.5).astype(int)

print("Predicted:", y_pred1[50][0])
print("Actual:", y_test[50])
```
