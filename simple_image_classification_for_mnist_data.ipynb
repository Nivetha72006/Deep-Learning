```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
mnist=keras.datasets.mnist
(x_train,y_train),(x_test,y_test)=mnist.load_data()# here train and test images are already in the dataset
x_train,x_test=x_train/255.0,x_test/255.0 # here color ranges from 0 to 255. so we used 255.0 to normalize btw 0 to 1 range or to  make it gray scaled
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[40000])
```

```python
x_train.shape
```

```python
y_train[112]
```

```python
model=keras.Sequential([
    layers.Flatten(input_shape=(28,28,1)),#28x28 is 784
    layers.Dense(128,activation='relu'),# we cannot give greater number so we give an avg number simply to start
    layers.Dense(10,activation='softmax')
])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
hist=model.fit(x_train,y_train,validation_split=0.1,epochs=15)
```

```python
y_pred=model.predict(x_test)
```
